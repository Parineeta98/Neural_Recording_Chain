```python
import numpy as np
import matplotlib.pyplot as plt
from neural_recording_chain.frozen_noise import make_frozen_noise

dt_ms = 0.05
input_theory, hidden_state = make_frozen_noise(duration_ms=4000, dt_ms=dt_ms, seed=0)
time_ms = np.arange(len(input_theory)) * dt_ms
input_current = 20 * input_theory

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(time_ms, input_current, color="darkgrey", label="Input current")
ax.set_xlabel("Time (ms)")
ax.set_ylabel("Input current (uA)")

ax2 = ax.twinx()
ax2.plot(time_ms, hidden_state, color="black", label="Hidden state")
ax2.set_ylabel("Hidden state")

ax.set_title("Frozen noise input")
fig.legend(loc="lower right")
plt.show()
```
